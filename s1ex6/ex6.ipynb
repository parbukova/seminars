```python
f = open('input.txt')
numbers = f.readline()
sign = f.readline()
base1 = f.readline()
f.close()
base = int(base1)
plus = '+'
minus = '-'
comp = '*'
fin = ''
carry = 0
A = list(numbers.split())
first = A[0] 
second = A[1]  
length = max(len(first), len(second))
if sign == plus:
   one = first
   two = second
   if length > len(first):
      zero = '0'
      p = length - len(first)
      one = str(zero*p + first)
   if length > len(second):
      zero = '0'
      q = length - len(second)
      two = zero*q + second
   for i in range(length - 1, -1, -1):
      one_str = one[i]
      one_int = int(one[i])
      two_str = two[i]
      two_int = int(two[i])
      res1 = one_int + two_int + carry
      carry = 0
      res2 = res1
      if res1 > base:
         carry = 1
         res2 = res1 - base
      fin = str(res2) + fin
   if carry > 0:
      fin = str(carry) + fin
      
elif sign == minus:
   carry =0
   if length > len(first):
      zero = '0'
      p = length - len(first)
      one = str(zero*p + first)
   if length > len(second):
      zero = '0'
      q = length - len(second)
      two = zero*q + second
   for i in range(length - 1, -1, -1):
      one_str = one[i]
      one_int = int(one[i]) - carry
      carry = 0
      two_str = two[i]
      two_int = int(two[i])
      if one_int < two_int:
         carry = 1
         one_int += base
   res3 = one_int - two_int 
   fin = str(res3) + fin
else:
   list1 = [int(i) for i in first[::-1]]
   list2 = [int(j) for j in second[::-1]]
   result = [0] * (len(first)+len(second))
   for i in range(len(list1)):
       carry = 0
       one = list1[i]
       for j in range(len(list2)):
          two = list2[j]
          res4 = one*two + result[i+j] + carry
          carry = res4//base
          result[i+j] = res4%base
          if res4 > base:
            carry = res4//base
       result.append(res4)
       if carry > 0:
         result[i + len(list2)] += carry
   while len(result) > 1 and result[-1] == 0:
      result.pop()
   fin2 = result[::-1]
   fin = ''.join(str(i) for i in fin)
print(fin2)

g = open('output.txt', 'w')
g.write(str(fin))
g.close()
```
