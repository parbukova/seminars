задание 2

```python
n = int(input())

def factorisation(n):
   i = 2
   numbers = []
   while i * i <= n:
       while n % i == 0:
           numbers.append(i)
           n = n // i
       i = i + 1
   if n > 1:
       numbers.append(n)
   return numbers
print(factorisation(n))
```

```python
n = list(input().split())
size = int(n[0])
symb = n[1]
def func(size, symb):
    for i in range(0, ((size//2)+1)):
        print(symb*i)
    for i in range(((size//2)+1), 0, -1):
        print(symb*i)
    return ''

print(func(size, symb))

```
